```python
import pandas as pd
import numpy as np

watch_features = watch_history.groupby("user_id").agg(
    total_watch_time=("watch_duration_minutes", "sum"),
    avg_watch_progress=("progress_percentage", "mean"),
    num_sessions=("session_id", "count"),
    avg_rating_given=("user_rating", "mean")
).reset_index()

search_features = search_logs.groupby("user_id").agg(
    total_searches=("search_id", "count"),
    avg_results_returned=("results_returned", "mean"),
    avg_clicked_position=("clicked_result_position", "mean"),
    avg_search_duration=("search_duration_seconds", "mean"),
    typo_rate=("had_typo", "mean"),
    filter_usage_rate=("used_filters", "mean")
).reset_index()

rec_features = recommendation_logs.groupby("user_id").agg(
    total_recommendations=("recommendation_id", "count"),
    avg_recommendation_score=("recommendation_score", "mean"),
    click_through_rate=("was_clicked", "mean")
).reset_index()

review_features = reviews.groupby("user_id").agg(
    total_reviews=("review_id", "count"),
    avg_review_rating=("rating", "mean"),
    avg_sentiment_score=("sentiment_score", "mean"),
    verified_review_rate=("is_verified_watch", "mean")
).reset_index()

movie_ratings = watch_history.merge(movies[['show_id','final_rating']], left_on='movie_id', right_on='show_id', how='left')
movie_features = movie_ratings.groupby("user_id").agg(
    avg_movie_rating_watched=("final_rating", "mean"),
    num_unique_shows=("movie_id", "nunique")
).reset_index()

user_ml_data = users[['user_id','age','gender','subscription_plan','monthly_spend','is_active']].copy()

for df in [watch_features, search_features, rec_features, review_features, movie_features]:
    user_ml_data = user_ml_data.merge(df, on='user_id', how='left')

user_ml_data.fillna(0, inplace=True)

print("User-level ML-ready dataset:")
print(user_ml_data.head())
print("Shape:", user_ml_data.shape)

user_ml_data.to_csv("user_ml_data.csv", index=False)
```

Input Data:
We aggregated multiple behavioral datasets into a user-level dataset for the purpose of this demo. Each row represents a single user with features summarizing their interactions:

- Watch History: Total watch time, average progress, number of sessions, average ratings given
- Search Logs: Total searches, average results returned, average clicked position, search duration, typo rate, filter usage rate
- Recommendation Logs: Total recommendations received, average recommendation score, click-through rate
- Reviews: Total reviews, average review rating, average sentiment score, verified review rate
- Movies Watched: Average rating of watched movies, number of unique shows watched
- User Info: Age, gender, subscription plan, monthly spend, account active status

This dataset preserves the structural complexity of the full system, but is scaled down for the pupose of this demo.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import accuracy_score

user_ml_data = pd.read_csv("user_ml_data.csv")

user_ml_data['churn'] = (~user_ml_data['is_active']).astype(int)

features = [
    'age', 'monthly_spend', 'total_watch_time', 'num_sessions', 
    'avg_watch_progress', 'avg_rating_given', 'total_searches', 
    'avg_results_returned', 'avg_clicked_position', 'avg_search_duration',
    'typo_rate', 'filter_usage_rate', 'total_recommendations', 
    'avg_recommendation_score', 'click_through_rate', 'total_reviews', 
    'avg_review_rating', 'avg_sentiment_score', 'verified_review_rate', 
    'avg_movie_rating_watched', 'num_unique_shows', 'subscription_plan', 'gender'
]

X = user_ml_data[features]
y = user_ml_data['churn']

X = pd.get_dummies(X, columns=['subscription_plan', 'gender'], drop_first=True)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled, y, test_size=0.2, random_state=42
)

print("\n Scikit-learn Random Forest")
rf_model = RandomForestClassifier(n_estimators=50, random_state=42, n_jobs=-1)
rf_model.fit(X_train, y_train)

rf_preds = rf_model.predict(X_test)
rf_acc = accuracy_score(y_test, rf_preds)
print(f"Random Forest Test Accuracy: {rf_acc:.2f}")

importances = rf_model.feature_importances_
plt.figure(figsize=(10,6))
plt.barh(X.columns, importances)
plt.title("Feature Importance (Random Forest)")
plt.xlabel("Importance")
plt.ylabel("Features")
plt.show()

```

The user-level aggregated dataset consists of the foll0wing features:

- User demographics: age, monthly_spend, subscription_plan, gender
- Watch behavior: total_watch_time, num_sessions, avg_watch_progress, avg_rating_give
- Search behavior: total_searches, avg_results_returned, avg_clicked_position, avg_search_duration, typo_rate, filter_usage_rate
- Recommendations received: total_recommendations, avg_recommendation_score, click_through_rate
- Reviews written: total_reviews, avg_review_rating, avg_sentiment_score, verified_review_rate
- Movies watched: avg_movie_rating_watched, num_unique_shows

Target Variable: Churn (0=stay, 1=likely to cancel)

Output: Random Forest Test Accuracy: 0.86 (model correctly predicts 86% of churn cases).

The feaure importance graph shows how each feature contributes to the model's decisions. Example: avg_watch_progress is a strong predictor of churm, whereas, num_sessions has a moderate influence.

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

Input data: Same aggregated user-level dataset used for Random Forest.

Here numerical features were standardized to improve neural network training. In addition, Categorical variables were one-hot encoded (as in Random Forest).

```python
print(user_ml_data['churn'].value_counts())
```

```python
from sklearn.utils import class_weight
import numpy as np

class_weights = class_weight.compute_class_weight(
    class_weight='balanced',
    classes=np.unique(y_train),
    y=y_train
)

class_weights_dict = dict(enumerate(class_weights))

print("Class weights:", class_weights_dict)
```

Before training the neural network, we compute class weights to account for the imbalance in the target variable i.e., churn.
This way we ensure the model pays more attention to the minority class (churners).

Result: The weight for class 1 churn is higher because it is the minority class. These weights can be passed to TensorFlow's model.fit() to help the network focus on predicting churners.

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

tf_model = Sequential([
    Input(shape=(X_train_scaled.shape[1],)),
    Dense(32, activation='relu'),
    Dense(16, activation='relu'),
    Dense(1, activation='sigmoid')
])

tf_model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

Input layer: Matches the number of features after scaling/encoding.

Hidden layers: 32 and 16 neurons with relu activation capture non-linear patterns.

Output layer: Single neuron with sigmoid activation outputs churn probability.

Optimizer: for efficient gradient updates.
Loss function: binary classification purposes.
Metrics: tracks accuracy during training.


```python
history = tf_model.fit(
    X_train_scaled,
    y_train,
    epochs=30,
    batch_size=16,
    validation_split=0.2,
    class_weight=class_weights_dict,
    verbose=1
)
```

- Trains the model for 30 epochs with batches of 16 samples.
- Uses 20% of training data for validation.
- Applies class weights to account for churn imbalance.
- Outputs training progress for each epoch.

```python
loss, acc = tf_model.evaluate(X_test_scaled, y_test, verbose=0)
print(f"\nImproved Test Accuracy: {acc:.2f}")
```

Evaluates the trained model on the test set.

Output: Improved Test Accuracy: 0.58; shows how well the model predicts churn on unseen data. Lower than Scikit-learn because the model prioritizes churn probability detection rather than maximizing raw accuracy.

```python
probs = tf_model.predict(X_test_scaled)

print("First 10 churn probabilities:")
print(probs[:10])
```

Output: These probabilities indicate each user's risk of cancelling; with values closer to 1 indicating higher risk. This is particularly useful for targeted retention campaigns rather than just a binary yes/no prediction.

```python
tf_preds = (probs > 0.5).astype(int)

print("Binary predictions:")
print(tf_preds[:10])
```

Converts predicted probabilities into binary churn labels using a 0.5 threshold. Translates probabilities into a simple churn/no-churn prediction for comparison.

Observations compared to Scikit-lean Random Forest:
- Outputs probabilistic risk scores instead of direct feature importance.
- Requires feature scaling, class weighting, and careful hyperparameter tuning.
- Accuracy lower (0.58 vs 0.86), but model provides risk estimates that are valuable for business decisions.
- Less interpretable than Random Forest; hard to say which feature drives the prediction without additional analysis.
